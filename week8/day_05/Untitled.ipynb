KMeans Clustering:
- KMeans is a popular clusterong algorithms, which attempts to partition data into K clusters which is user-defined parameters.

-Objectives:
- To form clusters in such a way that samples falls into same cluster and dissimilar samples falls into different

KMeans Algorithm:

1.Initializing K centroids

2.Calculate the distance of each data point from each centroid.

3.Assign each data point to its closest centroid, creating a cluster

4.Recalculate the positions of the K centroids.

5.Repeat the steps 2-4 until centroids no longer move.
